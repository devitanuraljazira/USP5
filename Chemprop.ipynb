```python

import pandas as pd
import numpy as np
from pathlib import Path
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

# 1. Baca dataset USP5 yang digunakan untuk pelatihan
data_path = Path("/data/Zira/Dataset_USP5/USP5_chemprop_clean.csv")
pred_path = Path(
    "/data/Zira/Dataset_USP5/chemprop_v2_usp5_run1/model_0/test_predictions.csv"
)

data = pd.read_csv(data_path)
pred = pd.read_csv(pred_path)

# 2. Pisahkan nama kolom agar nilai aktual dan prediksi tidak tertukar
data = data[["SMILES", "pIC50"]].rename(
    columns={"pIC50": "pIC50_aktual"}
)

pred = pred[["SMILES", "pIC50"]].rename(
    columns={"pIC50": "pIC50_prediksi"}
)

# 3. Cocokkan berdasarkan SMILES
evaluasi = pred.merge(data, on="SMILES", how="left")

print("Jumlah senyawa uji:", len(evaluasi))
print("Nilai aktual yang ditemukan:", evaluasi["pIC50_aktual"].notna().sum())

display(evaluasi)

# 4. Hitung metrik hanya jika seluruh nilai aktual berhasil dicocokkan
if evaluasi["pIC50_aktual"].notna().all():
    aktual = evaluasi["pIC50_aktual"]
    prediksi = evaluasi["pIC50_prediksi"]

    rmse = np.sqrt(mean_squared_error(aktual, prediksi))
    mae = mean_absolute_error(aktual, prediksi)
    r2 = r2_score(aktual, prediksi)

    print("\nHASIL EVALUASI D-MPNN")
    print(f"RMSE : {rmse:.4f}")
    print(f"MAE  : {mae:.4f}")
    print(f"R²   : {r2:.4f}")
else:
    print(
        "\nAda SMILES yang belum cocok. "
        "Jangan menghitung metrik dahulu; periksa hasil tabel di atas."
    )
```

```python

import pandas as pd
import numpy as np
import json
from pathlib import Path
from sklearn.model_selection import KFold

# Dataset yang digunakan saat training
data_path = Path("/data/Zira/Dataset_USP5/USP5_chemprop_clean.csv")
cv_dir = Path("/data/Zira/Dataset_USP5/usp5_5fold_cv")
cv_dir.mkdir(parents=True, exist_ok=True)

data = pd.read_csv(data_path)

assert len(data) == 99, f"Jumlah data tidak sesuai: {len(data)}"
assert data["SMILES"].is_unique, "Masih ada SMILES duplikat"
assert data["pIC50"].notna().all(), "Ada pIC50 yang kosong"

# Membagi 99 senyawa menjadi 5 fold
kf = KFold(n_splits=5, shuffle=True, random_state=42)

splits = []

for fold, (train_idx, val_idx) in enumerate(kf.split(data), start=1):
    splits.append({
        "train": train_idx.tolist(),
        "val": val_idx.tolist(),
        "test": []
    })
    print(
        f"Fold {fold}: "
        f"training={len(train_idx)}, "
        f"validasi={len(val_idx)}"
    )

splits_path = cv_dir / "splits_5fold.json"

with open(splits_path, "w") as f:
    json.dump(splits, f, indent=2)

print("\nFile pembagian data tersimpan di:")
print(splits_path)
print("Total fold:", len(splits))
```

```python
import pandas as pd
import json
from pathlib import Path
from sklearn.model_selection import KFold, train_test_split

# Lokasi dataset dan folder output
data_path = Path(
    "/data/Zira/Dataset_USP5/USP5_chemprop_clean.csv"
)
cv_dir = Path(
    "/data/Zira/Dataset_USP5/usp5_5fold_cv"
)
cv_dir.mkdir(parents=True, exist_ok=True)

data = pd.read_csv(data_path)

assert len(data) == 99, f"Jumlah data tidak sesuai: {len(data)}"
assert data["SMILES"].is_unique, "Masih ada SMILES duplikat"
assert data["pIC50"].notna().all(), "Ada nilai pIC50 kosong"

# Outer 5-fold: menentukan data test setiap fold
outer_cv = KFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

splits = []

for fold, (remaining_idx, test_idx) in enumerate(
    outer_cv.split(data), start=1
):
    # Membagi data tersisa menjadi training dan validasi
    train_idx, val_idx = train_test_split(
        remaining_idx,
        test_size=0.2,
        random_state=42 + fold
    )

    split = {
        "train": train_idx.tolist(),
        "val": val_idx.tolist(),
        "test": test_idx.tolist()
    }
    splits.append(split)

    print(
        f"Fold {fold}: "
        f"training={len(train_idx)}, "
        f"validasi={len(val_idx)}, "
        f"test={len(test_idx)}"
    )

# Simpan file split baru
splits_path = cv_dir / "splits_5fold_nested.json"

with open(splits_path, "w") as f:
    json.dump(splits, f, indent=2)

print("\nFile tersimpan di:", splits_path)
print("Total fold:", len(splits))
```

```python

import pandas as pd
import numpy as np
import json

from pathlib import Path
from rdkit import Chem, DataStructs
from rdkit.Chem import rdFingerprintGenerator

# Lokasi file
data_path = Path("/data/Zira/Dataset_USP5/USP5_chemprop_clean.csv")
split_path = Path(
    "/data/Zira/Dataset_USP5/usp5_5fold_cv/splits_5fold_nested.json"
)

# Baca dataset dan pembagian fold
df = pd.read_csv(data_path)

with open(split_path, "r") as f:
    splits = json.load(f)

print("Ukuran dataset:", df.shape)
print("Kolom dataset:", df.columns.tolist())
print("\nContoh data:")
display(df.head())

# Pastikan kolom tersedia
assert "SMILES" in df.columns, "Kolom SMILES tidak ditemukan"
assert "pIC50" in df.columns, "Kolom pIC50 tidak ditemukan"

# Buat Morgan fingerprint radius 2, 2048 bit
generator = rdFingerprintGenerator.GetMorganGenerator(
    radius=2,
    fpSize=2048
)

fingerprints = []
invalid_rows = []

for i, smiles in enumerate(df["SMILES"]):
    mol = Chem.MolFromSmiles(smiles) if isinstance(smiles, str) else None

    if mol is None:
        invalid_rows.append(i)
        fingerprints.append(None)
    else:
        fingerprints.append(
            generator.GetFingerprintAsNumPy(mol)
        )

print("\nSMILES tidak valid:", len(invalid_rows))

if invalid_rows:
    print("Indeks SMILES tidak valid:", invalid_rows)
else:
    X = np.asarray(fingerprints, dtype=np.uint8)
    y = df["pIC50"].to_numpy(dtype=float)

    print("Bentuk fingerprint:", X.shape)
    print("Bentuk target:", y.shape)
    print("Jumlah fold:", len(splits))
```

```python

n = len(df)
all_indices = set(range(n))

for fold_no, split in enumerate(splits, start=1):
    train_idx = split["train"]
    val_idx = split["val"]
    test_idx = split["test"]

    # Memastikan tidak ada data yang tumpang tindih
    assert set(train_idx).isdisjoint(val_idx)
    assert set(train_idx).isdisjoint(test_idx)
    assert set(val_idx).isdisjoint(test_idx)

    # Memastikan seluruh indeks digunakan
    assert (
        set(train_idx) | set(val_idx) | set(test_idx)
    ) == all_indices

    print(
        f"Fold {fold_no}: "
        f"train={len(train_idx)}, "
        f"val={len(val_idx)}, "
        f"test={len(test_idx)}"
    )

print("\nSemua pembagian fold valid.")
```

```python
import pandas as pd
import numpy as np
from pathlib import Path
from sklearn.metrics import (
    mean_squared_error,
    mean_absolute_error,
    r2_score
)

base_dir = Path(
    "/data/Zira/Dataset_USP5/usp5_5fold_cv"
)

files = {
    "D-MPNN": base_dir / "chemprop_5fold_model"
        / "usp5_5fold_oof_predictions.csv",
    "RF baseline": base_dir / "RF_baseline"
        / "RF_5fold_predictions.csv",
    "RF optimized": base_dir / "RF_optimization"
        / "RF_optimized_oof_predictions.csv",
    "SVR baseline": base_dir / "SVR_baseline"
        / "SVR_5fold_predictions.csv",
    "SVR optimized": base_dir / "SVR_optimization"
        / "SVR_optimized_oof_predictions.csv",
}

hasil = []

for nama_model, path in files.items():
    df = pd.read_csv(path)

    kolom_fold = "fold" if "fold" in df.columns else "Fold"

    for fold, data_fold in df.groupby(kolom_fold):
        aktual = data_fold["pIC50_aktual"]
        prediksi = data_fold["pIC50_prediksi"]

        hasil.append({
            "Model": nama_model,
            "Fold": fold,
            "N": len(data_fold),
            "RMSE": np.sqrt(
                mean_squared_error(aktual, prediksi)
            ),
            "MAE": mean_absolute_error(aktual, prediksi),
            "R2": r2_score(aktual, prediksi)
        })

hasil_fold = pd.DataFrame(hasil)

display(
    hasil_fold.round(4)
    .sort_values(["Model", "Fold"])
    .reset_index(drop=True)
)
```
